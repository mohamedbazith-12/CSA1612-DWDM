```r
age <- c("1-5", "5-15", "15-20", "20-50", "50-80", "80-110")
frequency <- c(200, 450, 300, 1500, 700, 44)
N <- sum(frequency)
median_position <- N / 2
cumulative_frequency <- cumsum(frequency)
median_class_index <- which(cumulative_frequency >= median_position)[1]
lower_limit <- c(1, 5, 15, 20, 50, 80)[median_class_index]
class_frequency <- frequency[median_class_index]
previous_cumulative_frequency <- if (median_class_index == 1) {
  0
} else {
  cumulative_frequency[median_class_index - 1]
}
class_width <- c(4, 10, 5, 30, 30, 30)[median_class_index]
median_value <- lower_limit +
  ((median_position - previous_cumulative_frequency) /
  class_frequency) * class_width
cat("Total Frequency:", N, "\n")
cat("Median Position:", median_position, "\n")
cat("Median Class:", age[median_class_index], "\n")
cat("Approximate Median:", median_value, "\n")
```

```r
age <- c(13, 15, 16, 16, 19, 20, 20, 21, 22, 22,
         25, 25, 25, 25, 30, 33, 33, 35, 35, 35,
         35, 36, 40, 45, 46, 52, 70)
cat("Mean:", mean(age), "\n")
cat("Median:", median(age), "\n")
frequency <- table(age)
max_frequency <- max(frequency)
modes <- names(frequency[frequency == max_frequency])
cat("Mode:", paste(modes, collapse = ", "), "\n")
cat("Maximum frequency:", max_frequency, "\n")
if (length(modes) == 1) {
  cat("Modality: Unimodal\n")
} else if (length(modes) == 2) {
  cat("Modality: Bimodal\n")
} else {
  cat("Modality:", length(modes), "modes\n")
}
midrange <- (min(age) + max(age)) / 2
cat("Midrange:", midrange, "\n")
cat("Q1:", unname(quantile(age, 0.25)), "\n")
cat("Q3:", unname(quantile(age, 0.75)), "\n")
```

```r
x <- c(200, 300, 400, 600, 1000)
min_max <- (x - min(x)) / (max(x) - min(x))
cat("Original Data:\n")
print(x)
cat("\nMin-Max Normalization:\n")
print(round(min_max, 4))
z_score <- (x - mean(x)) / sd(x)
cat("\nMean:", mean(x), "\n")
cat("Standard Deviation:", sd(x), "\n")
cat("Z-Score Normalization:\n")
print(round(z_score, 4))
```

```r
data <- c(11, 13, 13, 15, 15, 16, 19, 20,
          20, 20, 21, 21, 22, 23, 24, 30,
          40, 45, 45, 45, 71, 72, 73, 75)
data <- sort(data)
bins <- split(data, ceiling(seq_along(data) / 3))
cat("Original Data:\n")
print(data)
bin_mean <- lapply(bins, function(b) {
  rep(mean(b), length(b))
})
cat("\n(a) Smoothing by Bin Mean:\n")
print(unlist(bin_mean))
bin_median <- lapply(bins, function(b) {
  rep(median(b), length(b))
})
cat("\n(b) Smoothing by Bin Median:\n")
print(unlist(bin_median))
bin_boundaries <- lapply(bins, function(b) {
  low <- min(b)
  high <- max(b)
  sapply(b, function(value) {
    if (abs(value - low) <= abs(value - high)) {
      low
    } else {
      high
    }
  })
})
cat("\n(c) Smoothing by Bin Boundaries:\n")
print(unlist(bin_boundaries))
```

```r
marks <- c(55, 60, 71, 63, 55, 65, 50, 55, 58,
           59, 61, 63, 65, 67, 71, 72, 75)
marks <- sort(marks)
equal_freq_bins <- split(marks, ceiling(seq_along(marks) / 6))
cat("Equal-Frequency Bins:\n")
print(equal_freq_bins)
breaks <- seq(min(marks), max(marks), length.out = 4)
equal_width_bins <- split(
  marks,
  cut(marks, breaks = breaks, include.lowest = TRUE,
      labels = FALSE)
)
cat("\nEqual-Width Bins:\n")
print(equal_width_bins)
hist(marks,
     breaks = breaks,
     main = "Histogram of Student Marks",
     xlab = "Marks",
     ylab = "Frequency",
     col = "lightblue",
     right = TRUE)
```
